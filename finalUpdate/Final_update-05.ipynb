```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df_main = pd.read_csv(r'FINAL_BINARY_DATASET.csv')
```

```python
target_mapping = {'No_Disease':0,
                  'Disease':1}
df_main['Target'] = df_main['Target'].map(target_mapping)
```

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier


selected_features = [
    'Age of the patient', 'Specific gravity of urine', 'Random blood glucose level (mg/dl)',
    'Blood urea (mg/dl)', 'Serum creatinine (mg/dl)', 'Sodium level (mEq/L)', 'Potassium level (mEq/L)',
    'Hemoglobin level (gms)',  'White blood cell count (cells/cumm)',
    'Red blood cell count (millions/cumm)', 'Estimated Glomerular Filtration Rate (eGFR)',
    'Urine protein-to-creatinine ratio', 'Urine output (ml/day)', 'Serum albumin level',
    'Parathyroid hormone (PTH) level', 'Serum calcium level', 'Duration of hypertension (years)',
    'Interleukin-6 (IL-6) level','Family history of chronic kidney disease','Serum phosphate level'
    ,'Pus cells in urine','Albumin in urine','Anemia (yes/no)'
]


df_main["Creatinine_eGFR_ratio"] = df_main["Serum creatinine (mg/dl)"] / df_main["Estimated Glomerular Filtration Rate (eGFR)"]

df_main["Age_hypertension_ratio"] = df_main["Age of the patient"] / (df_main["Duration of hypertension (years)"] + 1)
df_main["Serum_albumin_adjusted"] = df_main["Serum albumin level"] / (df_main["Urine output (ml/day)"] + 1)

df_main["total_blood_count"] = df_main["White blood cell count (cells/cumm)"] + df_main["Red blood cell count (millions/cumm)"]
df_main["electro_ratio"] = df_main["Sodium level (mEq/L)"] * df_main["Potassium level (mEq/L)"] **2
df_main["blood_creatinine_ratio"] = df_main["Blood urea (mg/dl)"] / df_main["Serum creatinine (mg/dl)"] 

df_main["Creatinine_eGFR_ratio"] = df_main["Serum creatinine (mg/dl)"] / df_main["Estimated Glomerular Filtration Rate (eGFR)"]
df_main["Age_hypertension_ratio"] = df_main["Age of the patient"] / (df_main["Duration of hypertension (years)"] + 1)
df_main["Serum_albumin_adjusted"] = df_main["Serum albumin level"] / (df_main["Urine output (ml/day)"] + 1)
df_main["total_blood_count"] = df_main["White blood cell count (cells/cumm)"] + df_main["Red blood cell count (millions/cumm)"]

df_main["blood_creatinine_ratio"] = df_main["Blood urea (mg/dl)"] / df_main["Serum creatinine (mg/dl)"]

df_main["eGFR_flag"] = (df_main["Estimated Glomerular Filtration Rate (eGFR)"] < 60).astype(int)
df_main["albumin_stress_index"] = df_main["Albumin in urine"] / (df_main["Serum creatinine (mg/dl)"] + 1)
df_main["hypertension_burden"] = df_main["Duration of hypertension (years)"] / (df_main["Age of the patient"] + 1)

df_main["electrolyte_imbalance"] = abs(df_main["Sodium level (mEq/L)"] - df_main["Potassium level (mEq/L)"])
df_main["blood_toxicity_score"] = df_main["Blood urea (mg/dl)"] * 0.6 + df_main["Serum creatinine (mg/dl)"] * 0.4

df_main["inflammation_ratio"] = df_main["Interleukin-6 (IL-6) level"] / (df_main["Serum albumin level"] + 1)


selected_features += [
    'Creatinine_eGFR_ratio', 'Age_hypertension_ratio', 'Serum_albumin_adjusted',
    'total_blood_count',  'blood_creatinine_ratio',
    'eGFR_flag', 'albumin_stress_index', 'hypertension_burden',
    'electrolyte_imbalance', 'blood_toxicity_score',
    'inflammation_ratio'
]



X = df_main[selected_features]
y = df_main["Target"]
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "SVM": SVC(),
    "KNN": KNeighborsClassifier(),
    "AdaBoost": AdaBoostClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric="logloss"),
    "LightGBM": LGBMClassifier(),
    "CatBoost": CatBoostClassifier(verbose=0)
}


for name, model in models.items():
    print(f"\n Evaluating: {name}")
    
    if name in ["Logistic Regression", "SVM", "KNN"]:
        model.fit(X_train_scaled, y_train)
        preds = model.predict(X_test_scaled)
    else:
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
    
    print(classification_report(y_test, preds))
```
