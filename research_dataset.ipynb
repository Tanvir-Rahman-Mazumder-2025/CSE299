```python
import pandas as pd

with open('chronic_kidney_disease_full.arff', 'r') as f:
    lines = f.readlines()

attributes = []
data_started = False
data = []

for line in lines:
    line = line.strip()
    if line.lower().startswith('@attribute'):
        attr_name = line.split()[1]
        attributes.append(attr_name)
    elif line.lower() == '@data':
        data_started = True
    elif data_started and line and not line.startswith('%'):
        values = [v.strip() if v.strip() != '' else None for v in line.split(',')]
        while len(values) < len(attributes):
            values.append(None)
        data.append(values[:len(attributes)])  # Truncate if too many values

df = pd.DataFrame(data, columns=attributes)
df.to_csv('ckd_cleaned.csv', index=False)

```

```python
# Step 2: Create DataFrame and clean column names
df = pd.DataFrame(data, columns=attributes)

# Strip quotes from column names
df.columns = [col.strip("'").strip('"') for col in df.columns]

df.to_csv('ckd_cleaned.csv', index=False)
```
